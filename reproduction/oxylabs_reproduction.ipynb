```python
import time, statistics
import requests
import pandas as pd
from bs4 import BeautifulSoup
from selenium import webdriver

URL = "https://sandbox.oxylabs.io/products"
N_RUNS = 3

def time_requests():
    t0 = time.perf_counter()
    r = requests.get(URL, headers={"User-Agent": "Mozilla/5.0"}, timeout=10)
    n = len(BeautifulSoup(r.text, "html.parser").select("div.product-card"))
    return {"total": time.perf_counter() - t0, "startup": 0.0, "cards": n}

def time_selenium():
    t0 = time.perf_counter()
    driver = webdriver.Chrome()
    t1 = time.perf_counter()
    try:
        driver.get(URL)
        html = driver.page_source
    finally:
        driver.quit()
    n = len(BeautifulSoup(html, "html.parser").select("div.product-card"))
    return {"total": time.perf_counter() - t0, "startup": t1 - t0, "cards": n}

rows = []
for method, fn in [("requests", time_requests), ("selenium", time_selenium)]:
    runs = [fn() for _ in range(N_RUNS)]
    rows.append({
        "method": method,
        "cards_found": runs[0]["cards"],
        "median_total_s": round(statistics.median(r["total"] for r in runs), 2),
        "median_startup_s": round(statistics.median(r["startup"] for r in runs), 2),
    })

timing = pd.DataFrame(rows)
timing["fetch_only_s"] = timing["median_total_s"] - timing["median_startup_s"]
timing
```

requests was ~4× faster per page (0.35s vs 1.42s, median of 3 runs); including startup, ~12×. Selenium is therefore unnecessary overhead for this site, and I used requests for pagination.

```python
import re
from urllib.parse import urljoin

BASE = "https://sandbox.oxylabs.io"

def text_or_none(parent, selector):
    el = parent.select_one(selector)
    return el.get_text(strip=True) if el else None

def parse_price(raw):
    if not raw:
        return None
    num = re.sub(r"[^\d,.]", "", raw).replace(".", "").replace(",", ".")
    try:
        return float(num)
    except ValueError:
        return None

def parse_card(card):
    link = card.select_one("a.card-header")
    href = link.get("href") if link else None
    last = href.rstrip("/").rsplit("/", 1)[-1] if href else ""
    price_raw = text_or_none(card, ".price-wrapper")
    rating_div = card.select_one("div.rating")
    return {
        "product_id": int(last) if last.isdigit() else None,
        "title": text_or_none(card, "h4.title"),
        "categories": "; ".join(s.get_text(strip=True) for s in card.select("p.category span")) or None,
        "rating": len(rating_div.select("svg")) if rating_div else None,
        "description": text_or_none(card, "p.description"),
        "price_raw": price_raw,
        "price_eur": parse_price(price_raw),
        "url": urljoin(BASE, href) if href else None,
    }

rows = [parse_card(c) for c in soup.select("div.product-card")]
df = pd.DataFrame(rows)
print(df.shape)
df.head()
```

**TURNING PAGES (10pages)**

```python
import requests
import pandas as pd
from bs4 import BeautifulSoup

all_rows = []

for page in range(1, 11):
    url = f"https://sandbox.oxylabs.io/products?page={page}"
    try:
        response = requests.get(url, timeout=10)
        page_soup = BeautifulSoup(response.text, "html.parser")
        cards = page_soup.select("div.product-card")
        for card in cards:
            row = parse_card(card)
            row["page"] = page
            all_rows.append(row)
    except requests.RequestException as e:
        print(f"page {page} failed: {e}")

df_all = pd.DataFrame(all_rows)
print(df_all.shape)
df_all.head()
```

Rating is 0 for every product, but page 1 scraped with Selenium earlier had ratings. 

```python
from selenium import webdriver

all_rows = []
driver = webdriver.Chrome()
try:
    for page in range(1, 11):
        driver.get(f"https://sandbox.oxylabs.io/products?page={page}")
        page_soup = BeautifulSoup(driver.page_source, "html.parser")
        for card in page_soup.select("div.product-card"):
            all_rows.append({**parse_card(card), "page": page})
finally:
    driver.quit()

df_all = pd.DataFrame(all_rows)
print(df_all.shape)
print(df_all["rating"].value_counts())
```

```python
df_all.to_csv("products_10pages.csv", index=False, encoding="utf-8-sig")
```

Ratings look right now. Saved the raw data first, then cleaned it below.

```python
df_clean = df_all.copy()

# 1. Duplicates
print("duplicate products:", df_clean.duplicated(subset="product_id").sum())
df_clean = df_clean.drop_duplicates(subset="product_id")

# 2. Line breaks and stray characters
df_clean["description"] = df_clean["description"].str.replace(r"\s+", " ", regex=True).str.strip()
df_clean["title"] = df_clean["title"].str.strip(" -")

# 3. Price conversion check
print("prices not converted:", df_clean["price_eur"].isna().sum())
df_clean = df_clean.drop(columns="price_raw")

# 4. Sanity check on ranges
print("price range:", df_clean["price_eur"].min(), "to", df_clean["price_eur"].max())
print("ratings found:", sorted(df_clean["rating"].unique()))

df_clean.head()
```

```python
df_clean.to_csv("products_10pages_clean.csv", index=False, encoding="utf-8-sig")
```
