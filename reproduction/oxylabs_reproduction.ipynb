**Finding the fields.** I printed one product card to see what's in it. Besides name and price, it has a link (with the product ID), categories, a rating and a description. Each element has two kinds of class names: readable ones like "title" and auto-generated ones like "css-7u5e79". 

```python
import time, statistics
import requests
import pandas as pd
from bs4 import BeautifulSoup
from selenium import webdriver

URL = "https://sandbox.oxylabs.io/products"
N_RUNS = 3

def time_requests():
    t0 = time.perf_counter()
    r = requests.get(URL, headers={"User-Agent": "Mozilla/5.0"}, timeout=10)
    n = len(BeautifulSoup(r.text, "html.parser").select("div.product-card"))
    return {"total": time.perf_counter() - t0, "startup": 0.0, "cards": n}

def time_selenium():
    t0 = time.perf_counter()
    driver = webdriver.Chrome()
    t1 = time.perf_counter()
    try:
        driver.get(URL)
        html = driver.page_source
    finally:
        driver.quit()
    n = len(BeautifulSoup(html, "html.parser").select("div.product-card"))
    return {"total": time.perf_counter() - t0, "startup": t1 - t0, "cards": n}

rows = []
for method, fn in [("requests", time_requests), ("selenium", time_selenium)]:
    runs = [fn() for _ in range(N_RUNS)]
    rows.append({
        "method": method,
        "cards_found": runs[0]["cards"],
        "median_total_s": round(statistics.median(r["total"] for r in runs), 2),
        "median_startup_s": round(statistics.median(r["startup"] for r in runs), 2),
    })

timing = pd.DataFrame(rows)
timing["fetch_only_s"] = timing["median_total_s"] - timing["median_startup_s"]
timing
```

requests found the same 32 product cards. The Selenium times changed between runs, so the exact number isn't reliable, but requests was faster every time. 

```python
import re
from urllib.parse import urljoin

BASE = "https://sandbox.oxylabs.io"

def text_or_none(parent, selector):
    el = parent.select_one(selector)
    return el.get_text(strip=True) if el else None

def parse_price(raw):
    if not raw:
        return None
    num = re.sub(r"[^\d,.]", "", raw).replace(".", "").replace(",", ".")
    try:
        return float(num)
    except ValueError:
        return None

def parse_card(card):
    link = card.select_one("a.card-header")
    href = link.get("href") if link else None
    last = href.rstrip("/").rsplit("/", 1)[-1] if href else ""
    price_raw = text_or_none(card, ".price-wrapper")
    rating_div = card.select_one("div.rating")
    return {
        "product_id": int(last) if last.isdigit() else None,
        "title": text_or_none(card, "h4.title"),
        "categories": "; ".join(s.get_text(strip=True) for s in card.select("p.category span")) or None,
        "rating": len(rating_div.select("svg")) if rating_div else None,
        "description": text_or_none(card, "p.description"),
        "price_raw": price_raw,
        "price_eur": parse_price(price_raw),
        "url": urljoin(BASE, href) if href else None,
    }

rows = [parse_card(c) for c in soup.select("div.product-card")]
df = pd.DataFrame(rows)
print(df.shape)
df.head()
```

**Improved extraction:** 
- 8 fields instead of 2. 
- The tutorial collects names and prices into two separate lists and joins them by position. Here each product card becomes one dict, so its fields always stay together. If an element is missing, the helper returns None instead of crashing. 
- Prices are converted to numbers ("91,99 €" becomes 91.99).

**TURNING PAGES (10pages)**

```python
import requests
import pandas as pd
from bs4 import BeautifulSoup

all_rows = []

for page in range(1, 11):
    url = f"https://sandbox.oxylabs.io/products?page={page}"
    try:
        response = requests.get(url, timeout=10)
        page_soup = BeautifulSoup(response.text, "html.parser")
        cards = page_soup.select("div.product-card")
        for card in cards:
            row = parse_card(card)
            row["page"] = page
            all_rows.append(row)
    except requests.RequestException as e:
        print(f"page {page} failed: {e}")

df_requests = pd.DataFrame(all_rows)
print(df_requests.shape)
df_requests.head()
```



The rating div is empty in the raw HTML, but in the browser it contains 5 star icons. The stars are drawn by JavaScript after the page loads, so requests never sees them. Comparing only the number of cards earlier wasn't enough: I should have compared the fields too. Switching pagination to Selenium, with one browser for all 10 pages.



```python
from selenium import webdriver

all_rows = []
driver = webdriver.Chrome()
try:
    for page in range(1, 11):
        driver.get(f"https://sandbox.oxylabs.io/products?page={page}")
        page_soup = BeautifulSoup(driver.page_source, "html.parser")
        for card in page_soup.select("div.product-card"):
            all_rows.append({**parse_card(card), "page": page})
finally:
    driver.quit()

df_all = pd.DataFrame(all_rows)
print(df_all.shape)
print(df_all["rating"].value_counts())
```

Ratings look right now

```python
df_all.to_csv("products_10pages.csv", index=False, encoding="utf-8-sig")
```

Saved the raw data first, then cleaned it below.

```python
df_clean = df_all.copy()

# 1. Duplicates
print("duplicate products:", df_clean.duplicated(subset="product_id").sum())
df_clean = df_clean.drop_duplicates(subset="product_id")

# 2. Line breaks and stray characters
df_clean["description"] = df_clean["description"].str.replace(r"\s+", " ", regex=True).str.strip()
df_clean["title"] = df_clean["title"].str.strip(" -")

# 3. Price conversion check
print("prices not converted:", df_clean["price_eur"].isna().sum())
df_clean = df_clean.drop(columns="price_raw")

# 4. Sanity check on ranges
print("price range:", df_clean["price_eur"].min(), "to", df_clean["price_eur"].max())
print("ratings found:", sorted(df_clean["rating"].unique()))

df_clean.head()
```

```python
df_clean.to_csv("products_10pages_clean.csv", index=False, encoding="utf-8-sig")
```

```python
import sqlite3

conn = sqlite3.connect("oxylabs.db")
df_clean.to_sql("products", conn, if_exists="replace", index=False)
print(pd.read_sql("SELECT COUNT(*) AS n_rows FROM products", conn))
conn.close()
```

## Summary

- **Reproduction:** got 32 product names and prices into a CSV. The blog title part no longer works because its class name 'e1dscegp1' has changed. Then use INSPECT to get the new class name 'e1ot91pl0' and works again.
- **requests vs Selenium:** requests was faster and found the same 32 products, but every rating came back as 0. The stars are drawn by JavaScript, so I switched to Selenium for turning 10 pages. 
- **Improvements:** 8 fields instead of 2; prices as numbers; one dict per product instead of parallel lists; 10 pages instead of 1.
- **Cleaning:** no duplicates, removed line breaks in descriptions, all prices converted, ratings checked against the live site.
- **Result:** 320 products saved to `products_10pages.csv` (raw) and `products_10pages_clean.csv` (cleaned).Saved the final data to a SQLite table.
